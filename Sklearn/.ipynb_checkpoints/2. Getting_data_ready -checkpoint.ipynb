## 1. Getting our data ready to be used with machine learning.
Three main things we have to do: 
1. Split the data into features and labels (usually X & y). 
2. Filling (also called imputing) or disregarding missing values.
3. Converting non-numerical values to numerical values (also called feature encoding).

```python
# 1. Get Data Ready.
heart_disease = pd.read_csv("../Pandas/Data/heart-disease.csv")
heart_disease.head()
```

```python
# Cerate X (feature matrix)
X = heart_disease.drop("target", axis=1)
X.head()
```

```python
# Create y (labels)
y = heart_disease["target"]
y.head()
```

```python
# Split the data into train and test data set.

from sklearn.model_selection import train_test_split
X_train, X_test,y_train, y_test= train_test_split(X,y,test_size=0.20)
```

## 1.1 Make sure it's all numerical

```python
car_sales = pd.read_csv("../Pandas/Data/car-sales-extended.csv")
car_sales
```

```python
len(car_sales)
```

```python
car_sales.dtypes
```

```python
# Split into X/y
X = car_sales.drop("Price", axis=1)
y = car_sales["Price"]

# Split into train and test
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2)
```

```python
# Build Machine Learning Model
from sklearn.ensemble import RandomForestRegressor
model = RandomForestRegressor()
model.fit(X_train,y_train)
model.fit(X_test,y_test)
```

```python
# Trun the catgories into numbers
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

catagorical_feature = ["Make","Colour", "Doors"]
one_hot = OneHotEncoder()
transformer = ColumnTransformer([("one_hot",
                                 one_hot,
                                 catagorical_feature)],
                               remainder = "passthrough")
transformed_X = transformer.fit_transfrom(X)
transformed_X
```
